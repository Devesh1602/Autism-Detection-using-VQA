```python
import numpy as np
import matplotlib.pyplot as plt

# Set width of bar
barWidth = 0.15
fig, ax = plt.subplots(figsize=(9, 4))

# Set height of bar
Threshold = [0.9808, 0.7800, 0.7080]
Mean = [0.9781, 0.7807, 0.6997]
Mode = [0.9770, 0.7723, 0.6982]

# Set position of bar on X axis for the first group
br1 = np.arange(len(Threshold))

# Set position of bar on X axis for the second group
br2 = [x + barWidth + 0.19 for x in br1]  # Adjusted position for gap between groups

# Set position of bar on X axis for the third group
br3 = [x + 2 * (barWidth + 0.19) for x in br1]  # Adjusted position for gap between groups


# Make the plot
plt.bar(br1, Threshold, color='maroon', width=barWidth,
        edgecolor='grey', label='Threshold')
plt.bar(br2, Mean, color='orange', width=barWidth,
        edgecolor='grey', label='Mean')
plt.bar(br3, Mode, width=barWidth,
        edgecolor='grey', label='Mode')

# Adding Xticks

plt.xticks([r + barWidth + 0.2 for r in range(len(Threshold))],
           ['VQAutism', 'ANN', 'Transformers'])

# Adding text labels on top of each bar
for r1, r2, r3, th, m, md in zip(br1, br2, br3, Threshold, Mean, Mode):
    ax.text(r1, th + 0.01, f'{th:.4f}', ha='center', va='bottom', color='black', fontsize=8)
    ax.text(r2, m + 0.01, f'{m:.4f}', ha='center', va='bottom', color='black', fontsize=8)
    ax.text(r3, md + 0.01, f'{md:.4f}', ha='center', va='bottom', color='black', fontsize=8)

plt.legend()
plt.show()
```
