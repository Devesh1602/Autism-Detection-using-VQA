## Code to find Mean of all the csv combined

```python
import pandas as pd

# List of columns to select
selected_columns = ['AU01', 'AU02', 'AU04', 'AU05', 'AU06', 'AU07', 'AU09', 'AU10', 'AU11', 'AU12', 'AU14', 'AU15', 'AU17', 'AU20', 'AU23', 'AU24', 'AU25', 'AU26', 'AU28', 'AU43']

# List of CSV file names
csv_files = ['video_features_1.csv','video_features_2.csv','video_features_3.csv','video_features_4.csv','video_features_5.csv','video_features_6.csv','video_features_7.csv','video_features_8.csv','video_features_9.csv','video_features_10.csv','video_features_11.csv','video_features_12.csv']

# Create an empty dictionary to store column means
column_means = {}

# Iterate over each CSV file
for file in csv_files:
    # Read the CSV file into a pandas DataFrame
    df = pd.read_csv(file)

    # Select the desired columns
    df = df[selected_columns]

    # Calculate the mean for selected columns
    means = df.mean()

    # Update the dictionary with column means
    for col, mean in zip(selected_columns, means):
        if col in column_means:
            column_means[col].append(mean)
        else:
            column_means[col] = [mean]

# Calculate the mean of column means
total_au_means = {}
for col, means in column_means.items():
    total_au_means[col] = sum(means) / len(means)

# Create a DataFrame from the total AU means
mean_df = pd.DataFrame({'AU': total_au_means.keys(), 'Total AU Mean': total_au_means.values()})

# Print the resulting mean DataFrame
mean_df
```

## Code to find Mode of all the csv combined

```python
import pandas as pd

# List of columns to select
selected_columns = ['AU01', 'AU02', 'AU04', 'AU05', 'AU06', 'AU07', 'AU09', 'AU10', 'AU11', 'AU12', 'AU14', 'AU15', 'AU17', 'AU20', 'AU23', 'AU24', 'AU25', 'AU26', 'AU28', 'AU43']

# List of CSV file names
csv_files = ['video_features_1.csv','video_features_2.csv','video_features_3.csv','video_features_4.csv','video_features_5.csv','video_features_6.csv','video_features_7.csv','video_features_8.csv','video_features_9.csv','video_features_10.csv','video_features_11.csv','video_features_12.csv']

# Create an empty dictionary to store column modes
column_modes = {}

# Iterate over each CSV file
for file in csv_files:
    # Read the CSV file into a pandas DataFrame
    df = pd.read_csv(file)

    # Calculate the mode for selected columns
    df = df[selected_columns]
    df = df.round(1)
    modes = df.mode().iloc[0]  # Retrieve the first row (mode) as a Series

    # Update the dictionary with column modes
    for col, mode in zip(selected_columns, modes):
        if col in column_modes:
            column_modes[col].append(mode)
        else:
            column_modes[col] = [mode]

# Calculate the mean of column modes
total_au_modes = {}
for col, modes in column_modes.items():
    total_au_modes[col] = sum(modes) / len(modes)

# Create a DataFrame from the total AU modes
mode_df = pd.DataFrame({'AU': total_au_modes.keys(), 'Total AU Mode': total_au_modes.values()})

# Print the resulting mode DataFrame
print(mode_df)
```
